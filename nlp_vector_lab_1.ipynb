```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df = pd.read_csv(r"C:\MCA_acadamics\Genai_updated\venv\Reviews.csv", nrows=50000)

print("Dataset Shape:", df.shape)
df.head()
```

Find Reviews Containing the Word "light"

```python
light_reviews = df[df["Text"].str.contains(r"\blight\b", case=False, na=False)]

print("Number of reviews containing 'light':", len(light_reviews))

light_reviews[["Text"]].head(20)
```

Select Three Sentences

Replace the example sentences below with actual reviews from your dataset that match the required contexts.

```python
sentences = [
    "This olive oil is wonderful and very light on the stomach.",
    
    "The packaging was broken and let too much light inside the box.",
    
    "I prefer this brand because it is a light and healthy snack."
]

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}:")
    print(sentence)
    print()
```

Create Bag-of-Words Vectors

```python
vectorizer = CountVectorizer()

dtm = vectorizer.fit_transform(sentences)

dense_matrix = dtm.toarray()

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nDocument-Term Matrix:")
print(dense_matrix)
```

Extract Individual Vectors

```python
vec_1 = dense_matrix[0]
vec_2 = dense_matrix[1]
vec_3 = dense_matrix[2]

print("Vector 1:", vec_1)
print("Vector 2:", vec_2)
print("Vector 3:", vec_3)
```

: Compute Dot Product Similarities

```python
similarity_1_2 = np.dot(vec_1, vec_2)

similarity_1_3 = np.dot(vec_1, vec_3)

print("Similarity (Sentence 1 vs Sentence 2):", similarity_1_2)
print("Similarity (Sentence 1 vs Sentence 3):", similarity_1_3)
```

Show Word Frequencies (Optional Visualization)

```python
feature_names = vectorizer.get_feature_names_out()

vector_df = pd.DataFrame(
    dense_matrix,
    columns=feature_names,
    index=["Sentence_1", "Sentence_2", "Sentence_3"]
)

vector_df
```
