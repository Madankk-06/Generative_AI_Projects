```python
import torch
import torch.nn.functional as F

##################################################
# Dummy Scores
##################################################

policy_chosen = torch.tensor([3.2])

policy_rejected = torch.tensor([1.4])

reference_chosen = torch.tensor([2.5])

reference_rejected = torch.tensor([2.1])

beta = 0.1

##################################################
# DPO Loss
##################################################

policy_difference = policy_chosen - policy_rejected

reference_difference = reference_chosen - reference_rejected

loss = -F.logsigmoid(

    beta*(

        policy_difference

        -

        reference_difference

    )

)

print("DPO Loss =",loss.item())
```
