```python
import autogen
from autogen_agentchat.agents import AssistantAgent

# -----------------------------
# Shared Blackboard (Memory)
# -----------------------------
blackboard = {}

# -----------------------------
# Agent 1: Code Generator
# -----------------------------
def code_generator(prompt):
    code = f"""
def greet():
    print("Hello! This code is generated for: {prompt}")
"""
    blackboard["code"] = code
    print("Code Generator completed.")
    return code

# -----------------------------
# Agent 2: System Auditor
# -----------------------------
def system_auditor():
    code = blackboard["code"]

    report = """
Audit Report:
✓ Code syntax looks correct.
✓ No security issues found.
✓ Function naming is appropriate.
"""

    blackboard["audit"] = report
    print("System Auditor completed.")
    return report

# -----------------------------
# Agent 3: QA Analyst
# -----------------------------
def qa_analyst():

    report = """
QA Report:
✓ Code passes quality checks.
✓ Ready for deployment.
"""

    blackboard["qa"] = report

    print("QA Analyst completed.")
    return report

# -----------------------------
# Main Workflow
# -----------------------------
user_prompt = input("Enter your task: ")

code_generator(user_prompt)

system_auditor()

qa_analyst()

# -----------------------------
# Final Output
# -----------------------------
print("\n========== FINAL RESULT ==========\n")

print("Generated Code:\n")
print(blackboard["code"])

print("\nAudit Report:\n")
print(blackboard["audit"])

print("\nQA Report:\n")
print(blackboard["qa"])
```
